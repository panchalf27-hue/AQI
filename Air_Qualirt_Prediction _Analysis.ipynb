```python
# STEP 13: AQI RANGE CHECK

print("\n========== AQI RANGE CHECK ==========")

print("Minimum AQI:", cleaned_df["india_aqi"].min())

print("Maximum AQI:", cleaned_df["india_aqi"].max())

print("Average AQI:", cleaned_df["india_aqi"].mean())

print(
    "Negative AQI Values:",
    (cleaned_df["india_aqi"] < 0).sum()
)
```

```python

# STEP 14: OUTLIER DETECTION USING IQR METHOD

print("\n========== OUTLIER DETECTION ==========")

columns_to_check = [
    "india_aqi",
    "pm2_5_mean",
    "pm10_mean",
    "no2_mean",
    "so2_mean",
    "o3_8h_max",
    "co_8h_max"
]

for column in columns_to_check:

    Q1 = cleaned_df[column].quantile(0.25)
    Q3 = cleaned_df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = cleaned_df[
        (cleaned_df[column] < lower_limit) |
        (cleaned_df[column] > upper_limit)
    ]

    print("\nColumn:", column)
    print("Lower Limit:", round(lower_limit, 2))
    print("Upper Limit:", round(upper_limit, 2))
    print("Outlier Count:", len(outliers))
```

```python

# STEP 15: OUTLIER PERCENTAGE ANALYSIS

print("\n========== OUTLIER PERCENTAGE ANALYSIS ==========")

for column in columns_to_check:

    Q1 = cleaned_df[column].quantile(0.25)
    Q3 = cleaned_df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outlier_count = (
        (cleaned_df[column] < lower_limit) |
        (cleaned_df[column] > upper_limit)
    ).sum()

    outlier_percentage = (
        outlier_count / len(cleaned_df)
    ) * 100

    print("\nColumn:", column)
    print("Total Outliers:", outlier_count)
    print("Outlier Percentage:", round(outlier_percentage, 2), "%")
```

```python

# STEP 16: NEGATIVE POLLUTION VALUES CHECK

print("\n========== NEGATIVE POLLUTION VALUES CHECK ==========")

pollution_columns = [
    "pm2_5_mean",
    "pm2_5_max",
    "pm10_mean",
    "pm10_max",
    "no2_mean",
    "so2_mean",
    "o3_8h_max",
    "co_8h_max",
    "dust_mean",
    "aod_mean"
]

for column in pollution_columns:

    negative_count = (cleaned_df[column] < 0).sum()

    print("\nColumn:", column)
    print("Negative Values:", negative_count)

print("\n========== NEGATIVE VALUE CHECK COMPLETED ==========")
```

```python

# STEP 17: CREATE NEXT-DAY AQI TARGET

print("\n========== NEXT-DAY AQI TARGET CREATION ==========")

# Create Next-Day AQI Target for Each City

cleaned_df["next_day_aqi"] = (
    cleaned_df.groupby("city")["india_aqi"].shift(-1)
)

# Check Missing Target Values

print(
    "Missing Next-Day AQI:",
    cleaned_df["next_day_aqi"].isnull().sum()
)

# Display Sample Records

print("\nSample Next-Day AQI Records:")

print(
    cleaned_df[
        ["date", "city", "india_aqi", "next_day_aqi"]
    ].head(10)
)

print("\nDataset Shape:", cleaned_df.shape)
```

```python

# STEP 18: REMOVE MISSING TARGET VALUES

print("\n========== REMOVE MISSING TARGET VALUES ==========")

# Check Missing Target Before Removal

print(
    "Missing Target Before:",
    cleaned_df["next_day_aqi"].isnull().sum()
)

# Remove Records with Missing Target

cleaned_df = cleaned_df.dropna(
    subset=["next_day_aqi"]
).copy()

# Convert Target Column to Integer

cleaned_df["next_day_aqi"] = (
    cleaned_df["next_day_aqi"].astype(int)
)

# Verify Final Dataset

print("\n========== FINAL DATASET VERIFICATION ==========")

print("Final Dataset Shape:", cleaned_df.shape)

print(
    "Missing Target After:",
    cleaned_df["next_day_aqi"].isnull().sum()
)

print("Total Cities:", cleaned_df["city"].nunique())

print("\nFirst 5 Records:")

print(
    cleaned_df[
        ["date", "city", "india_aqi", "next_day_aqi"]
    ].head()
)
```

```python

# STEP 19: FEATURE SELECTION

print("\n========== FEATURE SELECTION ==========")

# Select Input Features

selected_features = [
    "city",
    "india_aqi",
    "pm2_5_mean",
    "pm2_5_max",
    "pm10_mean",
    "pm10_max",
    "no2_mean",
    "so2_mean",
    "o3_8h_max",
    "co_8h_max",
    "dust_mean",
    "aod_mean",
    "temperature_2m_max",
    "temperature_2m_min",
    "temperature_2m_mean",
    "apparent_temperature_max",
    "precipitation_sum",
    "precipitation_hours",
    "relative_humidity_2m_mean",
    "wind_speed_10m_max",
    "wind_direction_10m_dominant",
    "shortwave_radiation_sum",
    "et0_fao_evapotranspiration",
    "weather_code"
]

# Select Target Column

target_column = "next_day_aqi"

# Display Selected Features

print("Total Input Features:", len(selected_features))

print("\nSelected Input Features:")
print(selected_features)

print("\nTarget Column:", target_column)

# Create Feature and Target DataFrames

X = cleaned_df[selected_features].copy()

y = cleaned_df[target_column].copy()

print("\nInput Features Shape:", X.shape)
print("Target Shape:", y.shape)

print("\nFeature Selection Completed!")
```

```python

# STEP 20: IDENTIFY CATEGORICAL AND NUMERICAL FEATURES

print("\n========== FEATURE DATA TYPES ==========")

# Check Data Types

print(X.dtypes)

# Identify Categorical Columns

categorical_columns = X.select_dtypes(
    include=["object"]
).columns.tolist()

# Identify Numerical Columns

numerical_columns = X.select_dtypes(
    include=["number"]
).columns.tolist()

print("\n========== CATEGORICAL FEATURES ==========")

print("Categorical Columns:", categorical_columns)

print("Total Categorical Features:", len(categorical_columns))

print("\n========== NUMERICAL FEATURES ==========")

print("Numerical Columns:", numerical_columns)

print("Total Numerical Features:", len(numerical_columns))

print("\nFeature Type Identification Completed!")
```

```python

# STEP 20: IDENTIFY CATEGORICAL AND NUMERICAL FEATURES

print("\n========== FEATURE DATA TYPES ==========")

print(X.dtypes)

# Identify Categorical Columns

categorical_columns = X.select_dtypes(
    include=["str", "object", "category"]
).columns.tolist()

# Identify Numerical Columns

numerical_columns = X.select_dtypes(
    include=["number"]
).columns.tolist()

print("\n========== CATEGORICAL FEATURES ==========")

print("Categorical Columns:", categorical_columns)
print("Total Categorical Features:", len(categorical_columns))

print("\n========== NUMERICAL FEATURES ==========")

print("Numerical Columns:", numerical_columns)
print("Total Numerical Features:", len(numerical_columns))

print("\nFeature Type Identification Completed!")
```

```python

# STEP 21: CITY CATEGORY ANALYSIS

print("\n========== CITY CATEGORY ANALYSIS ==========")

print("Total Unique Cities:", X["city"].nunique())

print("\nCity Names:")

print(sorted(X["city"].unique()))

print("\nRecords Per City:")

print(X["city"].value_counts().sort_index())

print("\nCity Category Analysis Completed!")
```

```python

# STEP 22: DATE FEATURE EXTRACTION

print("\n========== DATE FEATURE EXTRACTION ==========")

# Extract Month

X["month"] = cleaned_df["date"].dt.month

# Extract Day of Week

X["day_of_week"] = cleaned_df["date"].dt.dayofweek

# Extract Day of Year

X["day_of_year"] = cleaned_df["date"].dt.dayofyear

# Extract Year

X["year"] = cleaned_df["date"].dt.year


# Display Extracted Features

print("\nExtracted Date Features:")

print(
    X[
        ["month", "day_of_week", "day_of_year", "year"]
    ].head(10)
)

# Verify Dataset Shape

print("\nUpdated Input Features Shape:", X.shape)

print("\nDate Feature Extraction Completed!")
```

```python
X["day_of_week"] = cleaned_df["date"].dt.day_of_week

X["day_of_year"] = cleaned_df["date"].dt.day_of_year
```

```python
# STEP 23: FINAL FEATURE VERIFICATION

print("\n========== FINAL FEATURE VERIFICATION ==========")

# Display All Input Features
print("\nTotal Input Features:", X.shape[1])

print("\nFeature Names:")
print(X.columns.tolist())

# Check Missing Values
print("\nMissing Values in Input Features:")
print(X.isnull().sum().sum())

# Check Duplicate Records
print("\nDuplicate Records in Input Features:")
print(X.duplicated().sum())

# Check Target Variable
print("\nTarget Variable:", y.name)

print("\nTarget Shape:", y.shape)

# Final Dataset Shape
print("\nFinal X Shape:", X.shape)
print("Final y Shape:", y.shape)

print("\nFeature Verification Completed!")
```

```python
# STEP 24: CHRONOLOGICAL TRAIN-TEST SPLIT

print("\n========== CHRONOLOGICAL TRAIN-TEST SPLIT ==========")

# Get dates matching X records
dates = cleaned_df.loc[X.index, "date"]

# Get sorted unique dates
unique_dates = sorted(dates.unique())

# Calculate 80% date cutoff
split_index = int(len(unique_dates) * 0.80)

# Find cutoff date
cutoff_date = unique_dates[split_index]

print("\nTraining Data Cutoff Date:", cutoff_date)

# Create chronological masks
train_mask = dates < cutoff_date
test_mask = dates >= cutoff_date

# Split Input Features
X_train = X.loc[train_mask].copy()
X_test = X.loc[test_mask].copy()

# Split Target Variable
y_train = y.loc[train_mask].copy()
y_test = y.loc[test_mask].copy()

# Display Dataset Information
print("\n========== TRAINING DATA ==========")
print("X_train Shape:", X_train.shape)
print("y_train Shape:", y_train.shape)
print("Training Start Date:", dates.loc[train_mask].min())
print("Training End Date:", dates.loc[train_mask].max())

print("\n========== TESTING DATA ==========")
print("X_test Shape:", X_test.shape)
print("y_test Shape:", y_test.shape)
print("Testing Start Date:", dates.loc[test_mask].min())
print("Testing End Date:", dates.loc[test_mask].max())

print("\nTrain-Test Split Completed!")
```

```python
# STEP 25: IDENTIFY FEATURE TYPES

print("\n========== FEATURE TYPE IDENTIFICATION ==========")

# Identify Categorical Features
categorical_features = X_train.select_dtypes(
    include=["str", "object", "category"]
).columns.tolist()

# Identify Numerical Features
numerical_features = X_train.select_dtypes(
    include=["number"]
).columns.tolist()

print("\nCategorical Features:")
print(categorical_features)

print("\nTotal Categorical Features:", len(categorical_features))

print("\nNumerical Features:")
print(numerical_features)

print("\nTotal Numerical Features:", len(numerical_features))

print("\nTotal Features:",
      len(categorical_features) + len(numerical_features))

print("\nFeature Type Identification Completed!")
```

```python
# STEP 26: ONE-HOT ENCODING
%pip install scikit-learn
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer

print("\n========== ONE-HOT ENCODING ==========")

# Create One-Hot Encoder
city_encoder = OneHotEncoder(
    handle_unknown="ignore"
)

# Create Column Transformer
preprocessor = ColumnTransformer(
    transformers=[
        ("city", city_encoder, categorical_features)
    ],
    remainder="passthrough"
)

# Fit on Training Data and Transform
X_train_encoded = preprocessor.fit_transform(X_train)

# Transform Testing Data
X_test_encoded = preprocessor.transform(X_test)

# Display Results
print("\nOriginal Training Shape:", X_train.shape)
print("Encoded Training Shape:", X_train_encoded.shape)

print("\nOriginal Testing Shape:", X_test.shape)
print("Encoded Testing Shape:", X_test_encoded.shape)

print("\nOne-Hot Encoding Completed!")
```

```python
# STEP 27: ENCODED FEATURE VERIFICATION

print("\n========== ENCODED FEATURE VERIFICATION ==========")

# Get Encoded Feature Names
encoded_feature_names = preprocessor.get_feature_names_out()

print("\nTotal Encoded Features:", len(encoded_feature_names))

print("\nFirst 15 Encoded Feature Names:")
print(encoded_feature_names[:15])

print("\nLast 10 Encoded Feature Names:")
print(encoded_feature_names[-10:])

# Verify Dataset Shapes
print("\nTraining Encoded Shape:", X_train_encoded.shape)
print("Testing Encoded Shape:", X_test_encoded.shape)

print("\nEncoded Feature Verification Completed!")
```

```python
# STEP 28: IMPORT REGRESSION ALGORITHMS

from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.ensemble import GradientBoostingRegressor

print("\n========== ML ALGORITHMS ==========")

print("1. Linear Regression Imported")
print("2. Decision Tree Regressor Imported")
print("3. Random Forest Regressor Imported")
print("4. Gradient Boosting Regressor Imported")

print("\nAll ML Algorithms Imported Successfully!")
```

```python
# STEP 29: LINEAR REGRESSION MODEL TRAINING

print("\n========== LINEAR REGRESSION TRAINING ==========")

# Create Linear Regression Model
lr_model = LinearRegression()

# Train Model
lr_model.fit(X_train_encoded, y_train)

# Predict Testing Data
lr_predictions = lr_model.predict(X_test_encoded)

print("\nLinear Regression Model Trained Successfully!")

print("\nTotal Training Records:", len(X_train))
print("Total Testing Records:", len(X_test))

print("\nFirst 10 Actual AQI Values:")
print(y_test.head(10).tolist())

print("\nFirst 10 Predicted AQI Values:")
print(lr_predictions[:10])
```

```python
# STEP 30: DECISION TREE REGRESSION TRAINING

print("\n========== DECISION TREE TRAINING ==========")

# Create Decision Tree Model
dt_model = DecisionTreeRegressor(
    random_state=42,
    max_depth=15
)

# Train Model
dt_model.fit(X_train_encoded, y_train)

# Predict Testing Data
dt_predictions = dt_model.predict(X_test_encoded)

print("\nDecision Tree Model Trained Successfully!")

print("\nTotal Training Records:", len(X_train))
print("Total Testing Records:", len(X_test))

print("\nFirst 10 Actual AQI Values:")
print(y_test.head(10).tolist())

print("\nFirst 10 Predicted AQI Values:")
print(dt_predictions[:10])
```

```python
# STEP 31: RANDOM FOREST REGRESSION TRAINING

print("\n========== RANDOM FOREST TRAINING ==========")

# Create Random Forest Model
rf_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    max_depth=15,
    n_jobs=-1
)

# Train Model
rf_model.fit(X_train_encoded, y_train)

# Predict Testing Data
rf_predictions = rf_model.predict(X_test_encoded)

print("\nRandom Forest Model Trained Successfully!")

print("\nTotal Training Records:", len(X_train))
print("Total Testing Records:", len(X_test))

print("\nFirst 10 Actual AQI Values:")
print(y_test.head(10).tolist())

print("\nFirst 10 Predicted AQI Values:")
print(rf_predictions[:10])
```

```python

# STEP 32: GRADIENT BOOSTING REGRESSION TRAINING

print("\n========== GRADIENT BOOSTING TRAINING ==========")

# Create Gradient Boosting Model
gb_model = GradientBoostingRegressor(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=3,
    random_state=42
)

# Train Model
gb_model.fit(X_train_encoded, y_train)

# Predict Testing Data
gb_predictions = gb_model.predict(X_test_encoded)

print("\nGradient Boosting Model Trained Successfully!")

print("\nTotal Training Records:", len(X_train))
print("Total Testing Records:", len(X_test))

print("\nFirst 10 Actual AQI Values:")
print(y_test.head(10).tolist())

print("\nFirst 10 Predicted AQI Values:")
print(gb_predictions[:10])
```

```python
# STEP 33: MODEL PERFORMANCE EVALUATION

from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score

import numpy as np

print("\n========== MODEL PERFORMANCE EVALUATION ==========")

# Store All Model Predictions
model_predictions = {
    "Linear Regression": lr_predictions,
    "Decision Tree": dt_predictions,
    "Random Forest": rf_predictions,
    "Gradient Boosting": gb_predictions
}

# Evaluate Each Model
for model_name, predictions in model_predictions.items():

    mae = mean_absolute_error(y_test, predictions)

    rmse = np.sqrt(
        mean_squared_error(y_test, predictions)
    )

    r2 = r2_score(y_test, predictions)

    print("\nModel:", model_name)
    print("MAE:", round(mae, 2))
    print("RMSE:", round(rmse, 2))
    print("R2 Score:", round(r2, 4))

print("\nAll Models Evaluation Completed!")
```

```python
# STEP 34: MODEL COMPARISON TABLE

print("\n========== MODEL COMPARISON TABLE ==========")

import pandas as pd
import numpy as np

results = []

for model_name, predictions in model_predictions.items():

    mae = mean_absolute_error(y_test, predictions)

    rmse = np.sqrt(
        mean_squared_error(y_test, predictions)
    )

    r2 = r2_score(y_test, predictions)

    results.append({
        "Model": model_name,
        "MAE": round(mae, 2),
        "RMSE": round(rmse, 2),
        "R2 Score": round(r2, 4)
    })

# Create Comparison DataFrame
results_df = pd.DataFrame(results)

# Sort by MAE
results_df = results_df.sort_values(
    by="MAE"
).reset_index(drop=True)

print("\nModel Performance Ranking:")
print(results_df.to_string(index=False))

# Identify Best Model
best_model_name = results_df.iloc[0]["Model"]

print("\nBest Model Based on MAE:", best_model_name)

print("\nModel Comparison Completed!")
```

```python
# STEP 37: PREDICTION ERROR ANALYSIS

print("\n========== PREDICTION ERROR ANALYSIS ==========")

# Create Error DataFrame
error_df = pd.DataFrame({
    "Actual_AQI": y_test.values,
    "Predicted_AQI": gb_predictions
})

# Calculate Prediction Error
error_df["Error"] = (
    error_df["Actual_AQI"] -
    error_df["Predicted_AQI"]
)

# Calculate Absolute Error
error_df["Absolute_Error"] = (
    error_df["Error"].abs()
)

# Display First 10 Records
print("\nFirst 10 Prediction Errors:")
print(error_df.head(10))

# Error Statistics
print("\n========== ERROR STATISTICS ==========")

print("Mean Error:",
      round(error_df["Error"].mean(), 2))

print("Mean Absolute Error:",
      round(error_df["Absolute_Error"].mean(), 2))

print("Maximum Error:",
      round(error_df["Absolute_Error"].max(), 2))

print("Minimum Error:",
      round(error_df["Absolute_Error"].min(), 2))

print("\nPrediction Error Analysis Completed!")
```

```python
# STEP 40: BASELINE MODEL COMPARISON

print("\n========== BASELINE MODEL COMPARISON ==========")

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
import pandas as pd

# Baseline Prediction:
# Tomorrow's AQI = Today's AQI

baseline_predictions = X_test["india_aqi"].values

# Baseline Evaluation
baseline_mae = mean_absolute_error(
    y_test, baseline_predictions
)

baseline_rmse = np.sqrt(
    mean_squared_error(y_test, baseline_predictions)
)

baseline_r2 = r2_score(
    y_test, baseline_predictions
)

# Gradient Boosting Evaluation
gb_mae = mean_absolute_error(
    y_test, gb_predictions
)

gb_rmse = np.sqrt(
    mean_squared_error(y_test, gb_predictions)
)

gb_r2 = r2_score(
    y_test, gb_predictions
)

# Comparison Table
baseline_comparison = pd.DataFrame({
    "Model": ["Baseline Model", "Gradient Boosting"],
    "MAE": [baseline_mae, gb_mae],
    "RMSE": [baseline_rmse, gb_rmse],
    "R2 Score": [baseline_r2, gb_r2]
})

print("\nBaseline vs Gradient Boosting:")
print(baseline_comparison.round(4).to_string(index=False))

print("\nBaseline Model Comparison Completed!")
```
